```python
## Código corregido con codex

import re

archivo = "../data/baby1990.html"

with open(archivo, "r", encoding="utf-8") as f:
    contenido = f.read()

resultado = re.search(r"Popularity in (\d{4})", contenido)

if resultado:
    anio = resultado.group(1)
    print("Año:", anio)
```

Tarea 2: Extraer nombres y ranking -- Para extraer los nombres y sus rankings, revise la estructura de cada fila de la tabla, por eso use la expresión `<td>(\d+)</td><td>(\w+)</td><td>(\w+)</td>`

```python
## Para ver como está organizado el contenido de la tabla

for linea in contenido.splitlines():
    if "<td>" in linea:
        print(linea)
```

```python
patron = r"<td>(\d+)</td><td>(\w+)</td><td>(\w+)</td>"

coincidencias = re.findall(patron, contenido)

print(coincidencias[:10])
```

```python
nombres_ranking = []

for ranking, nombre_nino, nombre_nina in coincidencias:
    nombres_ranking.append((nombre_nino, ranking))
    nombres_ranking.append((nombre_nina, ranking))

print(nombres_ranking[:10])
```

Tarea 3: Guardar en un diccionario -- Cree el diccionario a partir de las lista creada en la tarea 2. Para este dccionario cada clave es un nombre y cada valor es el ranking. Adedmás con la expresion "if nombre not in diccionario" controla que un nombre solo aparezca una vez.

```python
## Uso la lista nombres_ranking que cree en la tarea 2

diccionario = {}

for nombre, ranking in nombres_ranking:
    if nombre not in diccionario:
        diccionario[nombre] = ranking

list(diccionario.items())[:10]
```

Tarea 4: Lista de nombre ordenadas alfabeticamente -- Use el diccionario creado en la tarea 3 y ordene alfabeticamente las claves.

```python
## Uso las claves del diccionario que son los nombres

nombres_ordenados = sorted(diccionario.keys())

print(nombres_ordenados)
```
